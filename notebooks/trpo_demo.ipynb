## What is TRPO?

Trust Region Policy Optimization (TRPO) is a policy gradient method that improves stability by constraining policy updates to stay within a "trust region". This prevents catastrophic performance drops that can occur with standard policy gradient methods.

### Key Features:
- Uses KL divergence constraint instead of clipping (like PPO)
- Employs conjugate gradient to solve optimization subproblem
- More stable but computationally heavier than PPO
- Excellent for continuous control tasks
